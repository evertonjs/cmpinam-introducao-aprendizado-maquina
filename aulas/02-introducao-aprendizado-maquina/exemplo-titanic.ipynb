```python
# Carregar a base de dados titanic da lib seaborn
import seaborn as sns
titanic = sns.load_dataset('titanic')
```

```python
# Mostrar os tipos de dados de cada coluna
titanic.dtypes
```

```python
import matplotlib.pyplot as plt

# Proporção de sobreviventes
proporcoes = titanic['survived'].value_counts(normalize=True).sort_index()

cores = ["#e74c3c", "#2ecc71"]
rotulos = ["Não sobreviveu (0)", "Sobreviveu (1)"]

fig, ax = plt.subplots(figsize=(7, 5))

bars = ax.bar(rotulos, proporcoes.values, color=cores, edgecolor="white", linewidth=0.8, width=0.5)

# Valor percentual em cada barra
for bar, val in zip(bars, proporcoes.values):
    ax.text(bar.get_x() + bar.get_width() / 2, val + 0.01,
            f"{val:.1%}", ha="center", va="bottom", fontsize=12, fontweight="bold")

ax.set_title("Proporção de sobreviventes no Titanic", fontsize=14, fontweight="bold", pad=12)
ax.set_ylabel("Proporção", fontsize=11)
ax.set_ylim(0, 1)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f"{y:.0%}"))
ax.grid(axis="y", linestyle="--", alpha=0.4)
ax.spines[["top", "right"]].set_visible(False)
ax.tick_params(axis="x", labelsize=11)

plt.tight_layout()
plt.show()

print(f"\nTotal de passageiros: {len(titanic)}")
print(proporcoes.rename({0: 'Não sobreviveu', 1: 'Sobreviveu'}).apply(lambda v: f"{v:.1%}").to_string())
```
