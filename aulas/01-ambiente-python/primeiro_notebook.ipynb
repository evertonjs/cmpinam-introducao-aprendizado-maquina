```python
# Função para calcular a media de uma lista de números
def calcular_media(numeros):
    if len(numeros) == 0:
        return 0
    return sum(numeros) / len(numeros)
```

```python
# Função para calcular o desvio padrão de uma lista de números
# usando a bliboteca numpy
import numpy as np
def calcular_desvio_padrao(numeros):
    if len(numeros) == 0:
        return 0
    return np.std(numeros)
```

```python
# Chamar a função com uma lista de números
numeros = [10, 20, 30, 40, 50]
media = calcular_media(numeros)
print("A média dos números é:", media)
```

```python
# Chamar a função para calcular o desvio padrão
desvio_padrao = calcular_desvio_padrao(numeros)
print("O desvio padrão dos números é:", desvio_padrao)
```

```python
# Diferença entre lista e tupla
# Lista é mutável, ou seja, pode ser alterada depois de criada
# Tupla é imutável, ou seja, não pode ser alterada depois de criada
lista = [1, 2, 3]
tupla = (1, 2, 3)
print("Lista:", lista)
print("Tupla:", tupla)
```

```python
# Adicionar um novo elemento à lista
lista.append(4)
print("Lista após adicionar um elemento:", lista)
```

```python
# Tentar adicionar um novo elemento à tupla (isso causará um erro)
try:
    tupla.append(4)
except AttributeError as e:
    print("Erro ao tentar adicionar um elemento à tupla:", e)
```

```python
# Criar uma classe para representar um carro
class Carro:
    def __init__(self, marca, modelo, ano):
        self.marca = marca
        self.modelo = modelo
        self.ano = ano

    def exibir_informacoes(self):
        print(f"Carro: {self.marca} {self.modelo} ({self.ano})")
```

```python
# Criar um objeto da classe Carro e exibir suas informações
meu_carro = Carro("Toyota", "Corolla", 2020)
meu_carro.exibir_informacoes()
```
