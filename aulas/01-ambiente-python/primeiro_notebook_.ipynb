```python
# Declaração de variáveis de diferentes tipos
inteiro = 42
flutuante = 3.14
texto = "Olá, mundo!"

# Exibindo os valores das variáveis
print("Valor do inteiro:", inteiro)
print("Valor do flutuante:", flutuante)
print("Valor do texto:", texto)
```

```python
# Estrutura de controle: Condicional
if inteiro > 10:
    print("O inteiro é maior que 10.")
else: print("O inteiro é menor ou igual a 10.")

# Estrutura de controle: Laço de repetição
for i in range(5):
    print("Contagem:", i)
```

```python
# Estrutura de controle com switch-case
def dia_da_semana(dia):
    switcher = {
        1: "Segunda-feira",
        2: "Terça-feira",
        3: "Quarta-feira",
        4: "Quinta-feira",
        5: "Sexta-feira",
        6: "Sábado",
        7: "Domingo"
    }
    return switcher.get(dia, "Dia inválido")
# Testando a função switch-case
print(dia_da_semana(3))  # Saída: Quarta-feira
```

```python
# Entrada de dados do usuário
nome = input("Digite seu nome: ")
print("Olá, " + nome + "! Bem-vindo ao curso de Python!")
```

```python
# Exemplo de dicionário
pessoa = {
    "nome": "Alice",
    "idade": 30,
    "cidade": "São Paulo",
    "hobbies": ["leitura", "viagem", "culinária"]
}
print("Informações da pessoa:", pessoa)
```

```python
# Criar um dicionário vazio e adicionar elementos
carro = {}
carro["marca"] = "Toyota"
carro["modelo"] = "Corolla"
carro["ano"] = 2020
print("Informações do carro:", carro)
```

```python
# Diferença entre listas e tuplas
lista = [1, 2, 3, 4, 5]  # Lista é mutável
tupla = (1, 2, 3, 4, 5)  # Tupla é imutável
print("Lista:", lista)
print("Tupla:", tupla)
```

```python
# Acessando elementos de uma lista e tupla
print("Primeiro elemento da lista:", lista[0])  # Saída: 1
print("Primeiro elemento da tupla:", tupla[0])  # Saída: 1
```

```python
# Adicionando elementos a uma lista
lista.append(6)
print("Lista após adicionar um elemento:", lista)
```

```python
# Tentar adicionar elementos a uma tupla (isso causará um erro)
try:
    tupla.append(6)  # Isso não é permitido, pois tuplas são imutáveis
except AttributeError as e:
    print("Erro ao tentar modificar a tupla:", e)
```
